### yfinance 사용

```python
import yfinance as yf

# Microsoft (MSFT)에 대한 Ticker 객체 생성
msft = yf.Ticker("MSFT")

# Ticker 객체에 대한 정보 출력 (.py에서 실행할 때는 print(msft.info)로 사용)
# display(msft.info)  
# 분석가들의 목표 주가 정보, 지금까지 3개월간 데이터로, 미래 12개월의 정보 예측
print(msft.info['targetLowPrice'])    # 최저 목표가
print(msft.info['targetMeanPrice'])   # 평균 목표가 (미래 기대치)
print(msft.info['targetHighPrice'])   # 최고 목표가
print(msft.info['numberOfAnalystOpinions']) # 참여한 분석가 수
```

```python
hist = msft.history(period="5d") # 5일간의 주가 데이터를 가져옴, 2개월:2mo, 1년:1y
display(hist) # 데이터 출력
```

```python
# recommandation :주식 분석가들이 이 주식에 대해 내린 추천（예 : 매수, 매도, 보유 등）을 포함하는 데이터프레임 반환
# period는 분석가들이 추천한 시점

msft.recommendations # 추천 정보 출력, 숫자는 주식분석가 의견수(몇명)
```

```python
msft.recommendations.plot(
    kind='bar', 
    title='Microsoft Analyst Recommendations', 
    xlabel='Period (Months)', 
    ylabel='Number of Analysts',
    figsize=(12, 6) # 가로 12인치, 세로 6인치로 설정
)
```
