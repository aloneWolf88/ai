```python
from dotenv import load_dotenv
import os

# openai key 가져오기
load_dotenv()
api_key = os.getenv('OPENAI_API_KEY')
```

```python
import json 

from openai import OpenAI

# OpenAI class 객체 생성
client = OpenAI(api_key= api_key) # OPENAI_API_KEY

prompt = '복싱의 효과 알려줘.';

response = client.chat.completions.create( 
    model="gpt-5.6",
    messages=[
        {
            'role': 'system',
            'content': '너는 의사야'
        },
        {
            'role': 'user',
            'content': prompt + '\n\n출력 형식: JSON으로 출력해줘.' 
        }
    ],
    n=1,             # 응답수
    max_completion_tokens=2024,  # 생성할 최대 토큰 수 제한
)
```

```python
print(type(response)) # <class 'openai.types.chat.chat_completion.ChatCompletion'>
print(response)
```

```python
print(type(response.choices[0].message.content))
```

```python
# print(response.choices[0].message.content['요가의_효과']) # X
```

```python
res = json.loads(response.choices[0].message.content) # str -> dict(json)
print(type(res)) # <class 'dict'>
display(res['복싱의_효과'])
```
